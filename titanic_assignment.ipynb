```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("train.csv")

df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nColumns:")
print(df.columns)

print("\nDataset Information:")
df.info()
```

```python
overall_survival_rate = df["Survived"].mean() * 100

print("Overall Survival Rate:", overall_survival_rate, "%")
```

```python
gender_distribution = df["Sex"].value_counts()

print("Gender Distribution:")
print(gender_distribution)
```

```python
gender_percentage = df["Sex"].value_counts(normalize=True) * 100

print("Gender Distribution Percentage:")
print(gender_percentage)
```

```python
survival_by_gender = df.groupby("Sex")["Survived"].mean() * 100

print("Survival Rate by Gender:")
print(survival_by_gender)
```

```python
difference = survival_by_gender.max() - survival_by_gender.min()

print("Difference in Survival Rate:", difference, "%")
```

```python
print(df["Age"].describe())
```

```python
df["Age"].plot(kind="hist", bins=20)

plt.xlabel("Age")
plt.ylabel("Number of Passengers")
plt.title("Age Distribution of Titanic Passengers")

plt.show()
```

```python
survival_by_class = df.groupby("Pclass")["Survived"].mean() * 100

print("Survival Rate by Passenger Class:")
print(survival_by_class)
```

```python
difference = survival_by_class.max() - survival_by_class.min()

print("Difference in Survival Rate:", difference, "%")
```
